```python
import boto3

service_quotas = boto3.client("service-quotas", region_name="ap-south-1")

quotas = service_quotas.list_service_quotas(
    ServiceCode="bedrock"
)["Quotas"]

for quota in quotas:
    name = quota["QuotaName"].lower()

    if any(term in name for term in ["request", "token", "inference"]):
        print(quota["QuotaName"])
        print("Value:", quota["Value"])
        print("Code:", quota["QuotaCode"])
        print()
```

```python
import boto3

region = "ap-south-1"

quotas_client = boto3.client(
    "service-quotas",
    region_name=region,
)

paginator = quotas_client.get_paginator("list_service_quotas")

for page in paginator.paginate(ServiceCode="bedrock"):
    for quota in page["Quotas"]:
        name = quota["QuotaName"]

        if "haiku" in name.lower() or "anthropic" in name.lower():
            print(f"{name}")
            print(f"Limit: {quota['Value']}")
            print(f"Quota code: {quota['QuotaCode']}")
            print("-" * 60)
```
