### JSON schema for tool functions
- JSON schema tells Claude what arguments your function expects and how to use it properly
- It has two main component:
    - top part for the name and the description  (tool spec)
    - the next schema part describes the functions arguments (input schema), with sample data.
    - json and json schema are not the same.
    e:g: {
  "toolSpec": {
    "name": "get_current_datetime",
    "description": "Returns the current local date and time formatted according to the provided Python strftime format string.",
    "inputSchema": {
      "json": {
        "type": "object",
        "properties": {
          "date_format": {
            "type": "string",
            "description": "Python strftime format string. Defaults to '%Y-%m-%d %H:%M:%S'."
          }
        },
        "required": []
      }
    }
  }
}
